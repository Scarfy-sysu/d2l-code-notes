# 📘 11.7 AdaGrad算法 - 学习笔记

## 📌 一、AdaGrad 是什么？

**AdaGrad（Adaptive Gradient）** 是一种优化算法，它根据每个参数的历史梯度信息，**自适应地调整学习率**。

其核心思想是：

> **更新频繁的参数 → 学习率降低**
> **更新不频繁的参数 → 学习率保持较大**

这样可以更有效地应对训练中的不同维度特征（尤其是稀疏特征）带来的挑战。

---

## ⚙️ 二、算法原理（直观解释）

### 🔢 更新公式：

给定第 \$t\$ 步的梯度 \$\boldsymbol{g}\_t\$，历史累计梯度平方为 \$\boldsymbol{s}\_t\$，AdaGrad 的更新方式如下：

$$
\boldsymbol{s}_t = \boldsymbol{s}_{t-1} + \boldsymbol{g}_t^2
$$

$$
\boldsymbol{\theta} \leftarrow \boldsymbol{\theta} - \frac{\eta}{\sqrt{\boldsymbol{s}_t + \epsilon}} \odot \boldsymbol{g}_t
$$

其中：

* \$\eta\$：初始学习率
* \$\epsilon\$：防止除以0的小常数（如 \$10^{-6}\$）
* \$\odot\$：表示逐元素相乘（element-wise）

### 🧠 理解：

* 梯度大的方向，\$\boldsymbol{s}\_t\$ 快速增大 → 学习率下降 → 抑制更新
* 梯度小的方向，\$\boldsymbol{s}\_t\$ 增长慢 → 学习率大 → 加快收敛

---

## 🧪 三、代码实现

### ✅ 手动实现（from scratch）：

```python
def adagrad(params, states, hyperparams):
    eps = 1e-6
    for p, s in zip(params, states):
        with torch.no_grad():
            s[:] += torch.square(p.grad)
            p[:] -= hyperparams['lr'] * p.grad / torch.sqrt(s + eps)
        p.grad.zero_()
```

### 🔧 初始化状态变量：

```python
def init_adagrad_states(feature_dim):
    return (torch.zeros((feature_dim, 1)), torch.zeros((1,)))
```

## 📊 四、与SGD对比

| 对比项    | SGD（小批量随机梯度下降） | AdaGrad（自适应梯度） |
| ------ | -------------- | -------------- |
| 学习率调整  | 固定             | 每个参数维度不同       |
| 历史信息利用 | 无              | 利用所有过去梯度       |
| 收敛速度   | 震荡，可能慢         | 更平稳，较快收敛       |
| 缺点     | 不稳定            | 学习率过早变得太小      |
| 适合场景   | 普通数据           | 稀疏数据（如NLP）     |

---

## 📝 五、总结要点

* AdaGrad 会为 **每个参数维度单独调整学习率**。
* 使用 **历史梯度平方和** 来调节更新幅度。
* 在一些任务中可能 **早期收敛，后期学习率过小**。
* 它是 RMSProp、Adam 等更高级优化算法的基础。
