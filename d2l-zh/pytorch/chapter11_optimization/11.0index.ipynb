# 优化算法
:label:`chap_optimization`

截止到目前，本书已经使用了许多优化算法来训练深度学习模型。优化算法使我们能够继续更新模型参数，并使损失函数的值最小化。这就像在训练集上评估一样。事实上，任何满足于将优化视为黑盒装置，以在简单的设置中最小化目标函数的人，都可能会知道存在着一系列此类“咒语”（名称如“SGD”和“Adam”）。

但是，为了做得更好，还需要更深入的知识。优化算法对于深度学习非常重要。一方面，训练复杂的深度学习模型可能需要数小时、几天甚至数周。优化算法的性能直接影响模型的训练效率。另一方面，了解不同优化算法的原则及其超参数的作用将使我们能够以有针对性的方式调整超参数，以提高深度学习模型的性能。

在本章中，我们深入探讨常见的深度学习优化算法。深度学习中出现的几乎所有优化问题都是*非凸*的。尽管如此，在*凸问题*背景下设计和分析算法是非常有启发性的。正是出于这个原因，本章包括了凸优化的入门，以及凸目标函数上非常简单的随机梯度下降算法的证明。

## 11.2 梯度下降

### 批量GD

$\theta \leftarrow \theta - \eta \nabla f(\theta)$

### 随机SGD

$\theta \leftarrow \theta - \eta \nabla f_i(\theta)$

### 小批量SGD

$\theta \leftarrow \theta - \eta \cdot \frac{1}{B} \sum_i \nabla f_i(\theta)$

---

## 11.3 动量法

### 指数加权移动平均

$v_t = \beta v_{t-1} + (1-\beta)g_t$

### 动量更新

$\theta \leftarrow \theta - \eta v_t$

## 11.6 AdaGrad

* 历史梯度平方和:
  $s_t = s_{t-1} + g_t^2$
* 更新:
  $\theta \leftarrow \theta - \frac{\eta}{\sqrt{s_t + \epsilon}} g_t$

问题：学习率下降过快，末期埋没

---

## 11.7 RMSProp

* 使用指数加权移动平均
  $s_t = \gamma s_{t-1} + (1 - \gamma) g_t^2$
* 保持学习率稳定

---

## 11.8 Adadelta

* 弃用全局学习率
* 分形:
  $\Delta \theta_t = - \frac{\sqrt{\Delta \theta_{t-1}^2 + \epsilon}}{\sqrt{s_t + \epsilon}} g_t$
* 全自适应，更加稳定

---

## 11.9 Adam

* 加入动量 + RMSProp
* 一阶矩度:
  $v_t = \beta_1 v_{t-1} + (1-\beta_1) g_t$
* 二阶矩度:
  $s_t = \beta_2 s_{t-1} + (1-\beta_2) g_t^2$
* 偏差修正
  $\hat{v}_t = v_t / (1 - \beta_1^t), \hat{s}_t = s_t / (1 - \beta_2^t)$
* 更新:
  $\theta \leftarrow \theta - \eta \cdot \frac{\hat{v}_t}{\sqrt{\hat{s}_t} + \epsilon}$

---

## 11.10 Yogi

* Adam 在非凝场景可能不收敛
* Yogi更新方式：

$$
s_t = s_{t-1} + (1 - \beta_2) \cdot \text{sgn}(g_t^2 - s_{t-1}) \cdot g_t^2
$$

* 更可控、更稳定，适合 BERT / Transformer
