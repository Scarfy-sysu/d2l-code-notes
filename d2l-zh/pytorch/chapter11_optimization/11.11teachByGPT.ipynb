# 📘 11.11 学习率调度器 - 学习笔记

## 📌 一、什么是学习率调度器？

在训练神经网络时，**学习率（learning rate）** 是影响收敛速度与最终性能的关键超参数。

> **学习率调度器** 是一种自动或手动调整学习率的策略，用于：
>
> * 加快收敛
> * 提升模型性能
> * 防止过拟合或震荡

---

## 🧠 二、为什么要使用学习率调度器？

* 训练初期：需要较大的学习率，加快收敛
* 训练中后期：适当减小学习率，有助于跳出震荡，稳定收敛至更优解
* 如果一直使用固定学习率，容易出现：

  * **震荡不收敛**
  * **收敛过慢**
  * **跳过最优点**

---

## 📊 三、常见调度策略

| 策略名称        | 思路                | 示例（下降规律）                             |
| ----------- | ----------------- | ------------------------------------ |
| **阶梯下降**    | 每隔一定 epoch，乘以衰减因子 | `0.1 → 0.01 → 0.001`                 |
| **多段下降**    | 在预设时间点骤降          | `epoch == 10, 20 时衰减`                |
| **指数衰减**    | 学习率按指数持续衰减        | \$\eta\_t = \eta\_0 \cdot \gamma^t\$ |
| **余弦退火**    | 学习率呈余弦波下降（平滑）     | 初期降得快，后期缓慢收尾                         |
| **预热 + 衰减** | 前几轮缓慢升高，再进入下降阶段   | 常用于 Transformer 和 BERT               |

## 🔧 四、手动实现：`d2l`中定义的学习率函数

### 示例：分段恒定学习率（如：epoch <= 2时为0.1，之后为0.01）

```python
def scheduler(epoch):
    if epoch < 2:
        return 1
    else:
        return 0.1
```

结合使用：

```python
d2l.plot([scheduler(epoch) for epoch in range(10)], 'epoch', 'lr')
```

输出：学习率随 epoch 变化的折线图

## ⚙️ 五、PyTorch 的学习率调度器（内置）

```python
from torch.optim.lr_scheduler import LambdaLR, StepLR, MultiStepLR, ExponentialLR
```

### 常用示例：

#### ✅ 1. StepLR（等间隔衰减）

```python
scheduler = torch.optim.lr_scheduler.StepLR(trainer, step_size=5, gamma=0.1)
```

每 5 个 epoch 把学习率乘以 0.1。

#### ✅ 2. MultiStepLR（在多个时刻衰减）

```python
scheduler = torch.optim.lr_scheduler.MultiStepLR(trainer, milestones=[5, 10], gamma=0.1)
```

在第 5 和第 10 个 epoch 学习率衰减。

#### ✅ 3. LambdaLR（自定义函数）

```python
lambda_lr = lambda epoch: 0.95 ** epoch
scheduler = torch.optim.lr_scheduler.LambdaLR(trainer, lr_lambda=lambda_lr)
```
