# 目标检测数据集
:label:`sec_object-detection-dataset`

目标检测领域没有像MNIST和Fashion-MNIST那样的小数据集。
为了快速测试目标检测模型，[**我们收集并标记了一个小型数据集**]。
首先，我们拍摄了一组香蕉的照片，并生成了1000张不同角度和大小的香蕉图像。
然后，我们在一些背景图片的随机位置上放一张香蕉的图像。
最后，我们在图片上为这些香蕉标记了边界框。

## [**下载数据集**]

包含所有图像和CSV标签文件的香蕉检测数据集可以直接从互联网下载。


```python
%matplotlib inline
import os
import pandas as pd
import torch
import torchvision
from d2l import torch as d2l
```

```python
#@save
d2l.DATA_HUB['banana-detection'] = (
    d2l.DATA_URL + 'banana-detection.zip',
    '5de26c8fce5ccdea9f91267273464dc968d20d72')
```

## 读取数据集

通过`read_data_bananas`函数，我们[**读取香蕉检测数据集**]。
该数据集包括一个的CSV文件，内含目标类别标签和位于左上角和右下角的真实边界框坐标。


```python
#@save
def read_data_bananas(is_train=True):
    """读取香蕉检测数据集中的图像和标签"""
    data_dir = d2l.download_extract('banana-detection')
    csv_fname = os.path.join(data_dir, 'bananas_train' if is_train
                             else 'bananas_val', 'label.csv')
    csv_data = pd.read_csv(csv_fname)
    csv_data = csv_data.set_index('img_name')
    images, targets = [], []
    for img_name, target in csv_data.iterrows():
        images.append(torchvision.io.read_image(
            os.path.join(data_dir, 'bananas_train' if is_train else
                         'bananas_val', 'images', f'{img_name}')))
        # 这里的target包含（类别，左上角x，左上角y，右下角x，右下角y），
        # 其中所有图像都具有相同的香蕉类（索引为0）
        targets.append(list(target))
    return images, torch.tensor(targets).unsqueeze(1) / 256
```

### `read_data_bananas` 函数总结

**作用**：读取香蕉目标检测数据集中的图像和对应的标签（类别和边界框坐标）。

**输入参数**：
- `is_train=True`: 是否读取训练集（True）或验证集（False）

**输出内容**：
- `images`：一个图像列表，每个图像是 `(3, H, W)` 的张量
- `targets`：一个形状为 `(N, 1, 6)` 的张量，每行表示：
  `[类别索引, 左上角x, 左上角y, 右下角x, 右下角y]`，坐标归一化到 [0,1]

**说明**：
- 所有香蕉目标类别索引均为 `0`
- 图像和标签通过匹配文件名获取
- 用于目标检测模型的数据加载

通过使用`read_data_bananas`函数读取图像和标签，以下`BananasDataset`类别将允许我们[**创建一个自定义`Dataset`实例**]来加载香蕉检测数据集。


```python
#@save
class BananasDataset(torch.utils.data.Dataset):
    """一个用于加载香蕉检测数据集的自定义数据集"""
    def __init__(self, is_train):
        self.features, self.labels = read_data_bananas(is_train)
        print('read ' + str(len(self.features)) + (f' training examples' if
              is_train else f' validation examples'))

    # 输入参数：
    #     idx：数据集中的索引
    # 输出：
    #     返回一个元组 (image_tensor, label_tensor)
    #         image_tensor: 当前图像（转为 float 类型）
    #         label_tensor: 对应的标签信息（形状为 [1, 6]）
    def __getitem__(self, idx):
        return (self.features[idx].float(), self.labels[idx])

    # 返回整个数据集的样本数量
    def __len__(self):
        return len(self.features)
```

最后，我们定义`load_data_bananas`函数，来[**为训练集和测试集返回两个数据加载器实例**]。对于测试集，无须按随机顺序读取它。


```python
#@save
def load_data_bananas(batch_size):
    """加载香蕉检测数据集"""
    # shuffle=True 表示训练时会打乱数据顺序，避免模型记住数据顺序，提升泛化能力
    train_iter = torch.utils.data.DataLoader(BananasDataset(is_train=True),
                                             batch_size, shuffle=True)
    val_iter = torch.utils.data.DataLoader(BananasDataset(is_train=False),
                                           batch_size)
    return train_iter, val_iter
```

让我们[**读取一个小批量，并打印其中的图像和标签的形状**]。
图像的小批量的形状为（批量大小、通道数、高度、宽度），看起来很眼熟：它与我们之前图像分类任务中的相同。
标签的小批量的形状为（批量大小，$m$，5），其中$m$是数据集的任何图像中边界框可能出现的最大数量。

小批量计算虽然高效，但它要求每张图像含有相同数量的边界框，以便放在同一个批量中。
通常来说，图像可能拥有不同数量个边界框；因此，在达到$m$之前，边界框少于$m$的图像将被非法边界框填充。
这样，每个边界框的标签将被长度为5的数组表示。
数组中的第一个元素是边界框中对象的类别，其中-1表示用于填充的非法边界框。
数组的其余四个元素是边界框左上角和右下角的（$x$，$y$）坐标值（值域在0～1之间）。
对于香蕉数据集而言，由于每张图像上只有一个边界框，因此$m=1$。


```python
# 从香蕉检测数据集中读取一个小批量样本，并查看图像和标签的 shape（形状）。

batch_size, edge_size = 32, 256
# 设置 每个小批量 batch 的大小为 32
# 每张图像会被 统一缩放裁剪到 256×256 像素

train_iter, _ = load_data_bananas(batch_size)
# 调用前面定义好的 load_data_bananas 函数
# 返回训练集迭代器 train_iter，我们暂时不需要验证集，所以用 _ 忽略

batch = next(iter(train_iter))
# 这一步是核心：
#     把训练集迭代器 train_iter 转换成普通迭代器 iter(train_iter)
#     然后用 next() 取出 第一个 batch
# 得到的 batch 是一个包含两个元素的元组：
#     batch[0]：一个小批量的图像（32 张香蕉图）
#     batch[1]：对应的标签（每张图像的边框信息）
batch[0].shape, batch[1].shape
# 分别打印出图像和标签的形状
# batch[0]	[32, 3, 256, 256]	共 32 张图像，每张图像有 3 个通道（RGB），大小为 256×256
# batch[1]	[32, 1, 5]	        每张图像有 1 个边界框，每个边界框用 5 个数表示（如下）
```

## [**演示**]

让我们展示10幅带有真实边界框的图像。
我们可以看到在所有这些图像中香蕉的旋转角度、大小和位置都有所不同。
当然，这只是一个简单的人工数据集，实践中真实世界的数据集通常要复杂得多。


```python
imgs = (batch[0][0:10].permute(0, 2, 3, 1)) / 255
axes = d2l.show_images(imgs, 2, 5, scale=2)
for ax, label in zip(axes, batch[1][0:10]):
    d2l.show_bboxes(ax, [label[0][1:5] * edge_size], colors=['w'])
```

## 小结

* 我们收集的香蕉检测数据集可用于演示目标检测模型。
* 用于目标检测的数据加载与图像分类的数据加载类似。但是，在目标检测中，标签还包含真实边界框的信息，它不出现在图像分类中。
