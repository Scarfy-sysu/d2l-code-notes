## 12.1 编译器和解释器

### 一、本节核心解释

深度学习中，很多时候你写的是解释式Python代码，而环境中但需要编译执行的高效率代码。

**编译器 vs 解释器：**

* 解释器：按行执行Python代码，简单灵活
* 编译器：把代码转换成高效能的机器代码，速度快

在深度学习中，我们常应用编译器把解释式程序化为高效率执行代码，这种把Python代码"编译化"的技术称为**混合编程**（hybrid programming）。

---

### 二、关键概念

#### 12.1.1 符号变量 (symbolic variable)

* 符号变量是用于构造运算图的"空模型"
* 可以进行非实际计算，而是作为操作结构

#### 12.1.2 混合编程 (hybrid programming)

* 把Python函数编译成完整的运算图
* 通过 `@d2l.add_to_class` 和 `hybrid_forward` 实现
* 运行时无需Python采样化，速度更快

#### 12.1.3 Sequential的混合式编程

```python
from mxnet.gluon import nn

net = nn.HybridSequential()
net.add(nn.Dense(256, activation='relu'))
net.add(nn.Dense(2))
net.hybridize()  # 激活混合模式
```

---

### 三、总结

| 概念              | 作用               |
| --------------- | ---------------- |
| 编译器             | 将Python代码转为高效率代码 |
| 符号变量            | 允许构造可进行优化的运算图    |
| hybrid\_forward | 描述模型前向计算的逻辑      |
| hybridize()     | 把程序转为编译模式        |
